# Peer-Review Checklist — Pricing Memorandum Sign-off

Before a pricing memo produced by the agent is released to underwriting, a
peer reviewer must confirm:

- \[ \] Submission validation issues (if any) are disclosed, not silently fixed.
- \[ \] All trend, on-level, and credibility assumptions used trace to the
  Approved Trend Assumptions document (no unapproved substitutions).
- \[ \] Expense, capital and margin loads trace to the Expense and Capital
  Parameters document.
- \[ \] Every numeric result in the memo can be reproduced from the tool
  call trace (inputs → outputs at each step).
- \[ \] The memo clearly separates the Technical Price Indication (Section A)
  from the Commercial Decision (Section B), and Section B is left for
  underwriting management, not completed by the agent.
- \[ \] Large-loss / outlier treatment, if any sensitivity is shown, is labeled
  as a sensitivity and not silently baked into the base case without
  sign-off.
- \[ \] Sensitivities cover at minimum: attachment, limit, trend, and large-loss.
- \[ \] The memo states the comparison to the expiring price and rate change,
  without recommending a specific renewal price.
