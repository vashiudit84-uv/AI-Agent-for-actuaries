# Expense, Capital and Margin Parameters — Approved for Technical Pricing

## Expense Load

- Ceding company expense allowance: as disclosed in submission (brokerage +
  cedant expense), default 10.0% of premium if undisclosed.
- Internal treaty pricing/administration cost: 1.5% of premium (fixed).

## Cost of Capital

- Cat XL layers: capital charge is loaded as a percentage of premium equal to
  1.0× the layer’s 1-in-250-year modeled capital requirement multiplied by the
  hurdle rate (8.0% real, approved current rate).
- Simplified technical-pricing proxy (used when full capital model output is
  not attached to the submission): capital load = 12.0% of the blended
  expected loss cost. This proxy must be flagged in the memo as a simplification.

## Target Underwriting Margin

- Standard target technical margin for property cat XL: 6.0% of technical
  premium.
- This is a technical planning margin only; it is NOT the commercial margin
  underwriting may ultimately price to.

## Technical Price Formula

Technical Price = Blended Expected Loss Cost / (1 − Expense% − Capital Load% − Target Margin%)

All percentages above must come from this document or from submission-level
disclosures; the agent may not invent alternative loads.
