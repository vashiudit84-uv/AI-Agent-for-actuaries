# Approved Actuarial Assumptions — Effective Q3 2026

These are the ONLY assumptions the pricing agent may select without escalation.
Any assumption outside this list requires Chief Actuary sign-off and cannot be
applied automatically.

## Loss Trend (annual, compounding)

| Peril / Line | Approved Trend |
|----|---:|
| Property Cat — Wind | 6.0% |
| Property Cat — Earthquake | 3.5% |
| Property Cat — Severe Convective Storm | 8.0% |
| Property Cat — Wildfire | 7.5% |
| Property Cat — All-Peril Blend (default when peril split unavailable) | 6.5% |

## Exposure (TIV) Trend / On-Level

- Default TIV trend where not evidenced by submission data: 4.0% per annum.
- On-levelling for rate/limit profile changes must use submission-provided
  TIV by year where available; the default above is a fallback only and
  must be flagged as such in the memo.

## Premium On-Level

- Rate change on-levelling uses the cedant’s own historical rate-change
  disclosure. If undisclosed, premium is on-levelled using TIV trend only,
  and this substitution must be flagged.

## Credibility Standard

- Full credibility standard: 15 claims to the layer (property cat, low-frequency
  layer standard) per the Bühlmann-adjusted table on file.
- Partial credibility formula: Z = sqrt(min(1, n_claims / 15))
- Minimum data floor: fewer than 3 claims to layer over the experience period
  → Z is capped at 0.25 regardless of formula output.

## Large Loss / Outlier Treatment

- No loss may be excluded or capped from the experience period without a
  documented catastrophe-model-validated outlier justification reviewed by
  a peer reviewer. The agent may run a sensitivity showing the effect of
  removing a flagged large loss, but may not apply that adjustment to the
  base technical price indication without this sign-off.

## Prohibited Actions

- The agent must not substitute a trend, on-level factor, or credibility
  standard other than those listed above.
- The agent must not apply judgmental loading factors not defined in the
  Expense and Capital Parameters document.
