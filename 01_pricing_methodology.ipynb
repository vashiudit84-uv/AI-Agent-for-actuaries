# Property Catastrophe Treaty — Technical Pricing Methodology (v3.2)

## 1. Scope

Applies to all property catastrophe excess-of-loss (cat XL) treaty renewals and
new business submissions reviewed by the Treaty Pricing team.

## 2. Method Overview

The indicated technical price is built from two independent indications, blended
by credibility:

1.  **Experience (burning cost) indication** — historical losses to the layer,
    trended and on-levelled to the prospective period, divided by on-levelled
    subject premium.
2.  **Exposure (frequency/severity) indication** — a fitted frequency/severity
    model applied to current exposure (TIV, limit profile) to estimate the
    expected annual loss to the layer.

Blended Expected Loss Cost = Z × Experience Indication + (1 − Z) × Exposure Indication

## 3. Required Steps (in order)

1.  Validate submission data (completeness, consistency, min. history).
2.  Trend and on-level historical claims and premium.
3.  Apply the treaty structure (attachment, limit, reinstatements) to trended losses.
4.  Calculate the burning cost (annual and volume-weighted average).
5.  Fit frequency/severity models to exposure and loss data.
6.  Determine credibility weight Z per the credibility standard.
7.  Blend the two indications.
8.  Load for expenses, cost of capital and target margin to produce the
    **Technical Price Indication**.
9.  Run attachment, limit, trend, and large-loss sensitivities.
10. Compare to expiring price and prepare the pricing review note.

## 4. Minimum Data Requirements

- Minimum 10 years of loss history (5 years may be accepted with an explicit
  data-limitation flag and reduced credibility).
- Loss detail at the event/occurrence level, not only accident-year totals.
- Subject premium and TIV (total insured value) by year.
- Current-period exposure summary (TIV, occupancy/peril split where available).

## 5. Separation of Technical and Commercial Output

The technical price indication produced under this methodology is an
**actuarial reference point only**. It is not a quote, bind instruction, or
commercial recommendation. Commercial terms (final price, ceding commission,
brokerage load, competitive positioning) are determined separately by
underwriting management and are outside the scope of this methodology.
