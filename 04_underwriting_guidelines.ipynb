# Underwriting Guidelines — Property Catastrophe Treaty (extract relevant to pricing)

1.  Every renewal submission must be reviewed against expiring terms; any
    material change in attachment, limit, reinstatement provisions, or
    territory/peril scope must be explicitly called out in the pricing memo.
2.  The technical price indication must always be shown alongside the expiring
    technical price (where available) and the expiring risk-adjusted rate
    change, for underwriting context.
3.  Underwriting management, not the pricing analyst or the agent, holds
    authority to quote, bind, or set final commercial terms. The pricing memo
    is an input to that decision, not the decision itself.
4.  Any submission with fewer than the minimum required years of loss history,
    missing exposure detail, or unresolved data-quality issues must be flagged
    as “Indication Subject to Data Limitation” and cannot be presented as a
    clean technical price.
5.  Sensitivities are mandatory output for every memo: at minimum attachment,
    limit, trend, and large-loss sensitivities must be shown so underwriting
    can see the range around the point estimate.
